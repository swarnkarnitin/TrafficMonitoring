```python
import os
import shutil

# 1. Define the base paths for the source data (unzipped_vehicle_data) and the destination data (class_data).
base_source_path = "/content/drive/MyDrive/Traffic_Videos/roboflow_images/unzipped_vehicle_data"
base_destination_path = "/content/drive/MyDrive/Traffic_Videos/roboflow_images/class_data"

# 2. Define the list of subdirectories to process ('train', 'test', 'valid').
subdirectories_to_process = ['train', 'test', 'valid']

# 3. Define the list of class names ('bus', 'car', 'microbus', 'motorbike', 'pickup-van', 'truck').
class_names = ['bus', 'car', 'microbus', 'motorbike', 'pickup-van', 'truck']

# 4. Create a dictionary mapping class IDs (0 to 5, assuming they are zero-indexed based on the order of class_names) to class names.
class_id_to_name = {i: class_names[i] for i in range(len(class_names))}

# 5. Iterate through each subdirectory in subdirectories_to_process.
for subdirectory in subdirectories_to_process:
    current_source_subdirectory_path = os.path.join(base_source_path, subdirectory)

    # 6. Within each subdirectory, construct the paths to the 'images' and 'labels' directories.
    source_images_path = os.path.join(current_source_subdirectory_path, 'images')
    source_labels_path = os.path.join(current_source_subdirectory_path, 'labels')

    if os.path.exists(source_labels_path) and os.path.isdir(source_labels_path):
        # 7. Iterate through each file in the 'labels' directory.
        for filename in os.listdir(source_labels_path):
            if filename.endswith('.txt'):
                source_label_filepath = os.path.join(source_labels_path, filename)

                # 8. For each text file, open and read the first line to get the class ID.
                with open(source_label_filepath, 'r') as f:
                    first_line = f.readline().strip()
                    if not first_line:
                        continue # Skip empty lines

                    parts = first_line.split()
                    if not parts:
                        continue # Skip empty lines

                    try:
                        # 9. Convert the class ID to an integer.
                        class_id = int(parts[0])

                        # 10. Using the class ID to class name mapping, determine the destination class directory.
                        if class_id in class_id_to_name:
                            destination_class_name = class_id_to_name[class_id]
                            destination_class_dir = os.path.join(base_destination_path, destination_class_name)
                            destination_images_dir = os.path.join(destination_class_dir, 'images')
                            destination_labels_dir = os.path.join(destination_class_dir, 'labels')

                            # Construct the corresponding image filename
                            image_filename_base, _ = os.path.splitext(filename)
                            possible_image_extensions = ['.jpg', '.jpeg', '.png'] # Add other extensions if needed
                            source_image_filepath = None

                            for ext in possible_image_extensions:
                                temp_source_image_filepath = os.path.join(source_images_path, image_filename_base + ext)
                                if os.path.exists(temp_source_image_filepath):
                                    source_image_filepath = temp_source_image_filepath
                                    break

                            if source_image_filepath:
                                # 11. Construct the destination paths for the label file and the corresponding image file.
                                destination_label_filepath = os.path.join(destination_labels_dir, filename)
                                destination_image_filename = os.path.basename(source_image_filepath)
                                destination_image_filepath = os.path.join(destination_images_dir, destination_image_filename)

                                # 12. Copy the label file from the source 'labels' directory to the destination 'labels' directory.
                                shutil.copy2(source_label_filepath, destination_label_filepath)
                                # print(f"Copied label: {source_label_filepath} to {destination_label_filepath}") # Uncomment for debugging

                                # 13. Copy the image file from the source 'images' directory to the destination 'images' directory.
                                shutil.copy2(source_image_filepath, destination_image_filepath)
                                # print(f"Copied image: {source_image_filepath} to {destination_image_filepath}") # Uncomment for debugging
                            else:
                                print(f"Warning: Corresponding image not found for label file: {source_label_filepath}")

                        else:
                            print(f"Warning: Unknown class ID {class_id} found in file: {source_label_filepath}")

                    except ValueError:
                        print(f"Warning: Could not parse class ID from first value in file: {source_label_filepath}")

print("File copying process completed.")
```
