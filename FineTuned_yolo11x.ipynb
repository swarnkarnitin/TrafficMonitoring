```python
!wget "https://github.com/ultralytics/assets/releases/download/v8.3.0/yolo11n.pt" -O yolo11n.pt

from ultralytics import YOLO

# Load a model
model = YOLO('yolo11x.pt')  # load a pretrained model

# Train the model
results = model.train(data='data.yaml', epochs=20,
                         imgsz=640,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=0.0,
    translate=0.1,
    scale=0.5,
    shear=0.0,
    perspective=0.0,
    flipud=0.0,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.0,
    copy_paste=0.0,
    batch=4, # Further reduced batch size
    workers=8,
    cache=True,
    optimizer='AdamW'
)
```

```python
from ultralytics import YOLO
import os

# Load the fine-tuned model
model_path = '/content/runs/detect/train11/weights/best.pt'
model = YOLO(model_path)

# Create the output directory if it doesn't exist
output_dir = './content/output'
os.makedirs(output_dir, exist_ok=True)

# Run inference on the video for the first 1500 frames
video_path = '/content/MVI_0764.MP4'
results = model.predict(source=video_path, stream=True, save=True, save_dir=output_dir, max_det=1000, vid_stride=1, classes=range(10), conf=0.25, iou=0.45, show=False, save_conf=False, save_txt=False, exist_ok=True)

# Iterate through results to check for detections and potential issues
for i, r in enumerate(results):
    print(f"Processing frame {i+1}")
    if r.boxes:
        print(f"  Detections found: {len(r.boxes)}")
        # You can uncomment the line below to print bounding box coordinates for each detection
        # print(f"  Bounding boxes: {r.boxes.xyxy}")
    else:
        print("  No detections in this frame.")

print("Inference process finished.")
```
