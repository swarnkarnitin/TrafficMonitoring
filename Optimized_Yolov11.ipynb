## Run inference

### Subtask:
Run inference on the specified video file using the fine-tuned model.

**Reasoning**:
Load the fine-tuned model and run inference on the video file, limiting the processing to the first 100 frames.

```python
from ultralytics import YOLO
import cv2
import os

# Load the fine-tuned model
model = YOLO('/content/drive/MyDrive/Traffic_Videos/Trained_Model/best_yolo11n_50epc_v2.pt')

# Define the path to the video file
video_path = '/content/drive/MyDrive/Traffic_Videos/Kolkata_V2.mp4'

# Open the video file
cap = cv2.VideoCapture(video_path)

# Get video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the output directory for the processed frames or video
output_dir = '/content/inference_results'
os.makedirs(output_dir, exist_ok=True)

# Process the first 100 frames
frame_count = 0
while cap.isOpened() and frame_count < 100:
    ret, frame = cap.read()
    if not ret:
        break

    # Run inference on the current frame
    results = model(frame, stream=False, classes=[0, 1, 2, 3, 4, 5], conf=0.5, iou=0.7)

    # Save the annotated frame (optional)
    for r in results:
        im_array = r.plot()  # plot a BGR numpy array of predictions
        output_path = os.path.join(output_dir, f'frame_{frame_count:04d}.jpg')
        cv2.imwrite(output_path, im_array)

    frame_count += 1

# Release the video capture object
cap.release()

print(f"Inference on first {frame_count} frames completed. Results saved to {output_dir}")
```
