## Process each file

### Subtask:
For each file in the unzipped folder, read the data, apply the mapping to the first column, and save the modified data to the `out_tran_label` folder with the same filename.


**Reasoning**:
The next step is to iterate through the identified label files, read each one, apply the mapping, and save the modified file to the output directory.



```python
mapping = {0: 6, 1: 7, 2: 0, 3: 1, 4: 8, 5: 9, 6: 10, 7: 5, 8: 3}
output_dir = 'out_valid_label'

for filename in label_files:
    input_file_path = os.path.join('./in_valid_labels/labels/', filename)
    output_file_path = os.path.join(output_dir, filename)

    try:
        # Read the data
        df = pd.read_csv(input_file_path, sep=' ', header=None)

        # Apply the mapping to the first column
        df.loc[:, 0] = df.loc[:, 0].map(mapping)

        # Save the modified data
        df.to_csv(output_file_path, sep=' ', header=False, index=False)

        print(f"Processed '{filename}' and saved to '{output_file_path}'")

    except FileNotFoundError:
        print(f"Error: Input file '{input_file_path}' not found.")
    except Exception as e:
        print(f"An error occurred while processing '{filename}': {e}")
```

## Modify traffic signal labels

### Subtask:
Iterate through the label files in the unzipped `traffic_signal` dataset (train, test, and valid folders) and apply the provided mapping to the first column.


**Reasoning**:
Iterate through the label files in the traffic_signal dataset and apply the mapping.



```python
mapping = {0: 6, 1: 7, 2: 8}
traffic_signal_base_dir = "/content/drive/MyDrive/Traffic_Videos/traffic_signal"
splits = ["train", "test", "valid"]

for split in splits:
    labels_dir = os.path.join(traffic_signal_base_dir, split, "labels")
    if os.path.exists(labels_dir):
        print(f"Processing labels in: {labels_dir}")
        for filename in os.listdir(labels_dir):
            if filename.endswith(".txt"):
                file_path = os.path.join(labels_dir, filename)
                try:
                    df = pd.read_csv(file_path, sep=' ', header=None)
                    df.loc[:, 0] = df.loc[:, 0].map(mapping)
                    df.to_csv(file_path, sep=' ', header=False, index=False)
                    print(f"Processed and updated: {filename}")
                except FileNotFoundError:
                    print(f"Error: File not found - {file_path}")
                except Exception as e:
                    print(f"An error occurred while processing {filename}: {e}")
    else:
        print(f"Labels directory not found for split {split}: {labels_dir}")
```

```python
import os
import shutil

base_yolov3_dir = "/content/drive/MyDrive/Traffic_Videos/yolov3"
base_traffic_signal_dir = "/content/drive/MyDrive/Traffic_Videos/traffic_signal"
base_resultant_dir = "/content/drive/MyDrive/Traffic_Videos/resultant_dataset"

splits = ["train", "test", "valid"]
data_types = ["images", "labels"]

for split in splits:
    for data_type in data_types:
        source_yolov3_dir = os.path.join(base_yolov3_dir, split, data_type)
        source_traffic_signal_dir = os.path.join(base_traffic_signal_dir, split, data_type)
        destination_dir = os.path.join(base_resultant_dir, split, data_type)

        # Copy from yolov3
        if os.path.exists(source_yolov3_dir):
            print(f"Copying {data_type} from {source_yolov3_dir} to {destination_dir}")
            for filename in os.listdir(source_yolov3_dir):
                source_file = os.path.join(source_yolov3_dir, filename)
                destination_file = os.path.join(destination_dir, filename)
                shutil.copy2(source_file, destination_file)
        else:
            print(f"Source directory not found: {source_yolov3_dir}")

        # Copy from traffic_signal
        if os.path.exists(source_traffic_signal_dir):
            print(f"Copying {data_type} from {source_traffic_signal_dir} to {destination_dir}")
            for filename in os.listdir(source_traffic_signal_dir):
                source_file = os.path.join(source_traffic_signal_dir, filename)
                destination_file = os.path.join(destination_dir, filename)
                # Prevent overwriting if filenames are the same, you might want to handle this differently
                if not os.path.exists(destination_file):
                     shutil.copy2(source_file, destination_file)
                else:
                     print(f"Warning: Skipping copy for existing file {filename} in {destination_dir}")
        else:
            print(f"Source directory not found: {source_traffic_signal_dir}")

print("Merging complete.")
```

```python
import os
import pandas as pd

resultant_base_dir = "/content/drive/MyDrive/Traffic_Videos/resultant_dataset"
splits = ["train", "test", "valid"]

for split in splits:
    labels_dir = os.path.join(resultant_base_dir, split, "labels")
    if os.path.exists(labels_dir):
        print(f"Counting occurrences in labels for {split} split:")
        all_first_column_values = []
        for filename in os.listdir(labels_dir):
            if filename.endswith(".txt"):
                file_path = os.path.join(labels_dir, filename)
                try:
                    df = pd.read_csv(file_path, sep=' ', header=None)
                    if not df.empty:
                        all_first_column_values.extend(df.iloc[:, 0].tolist())
                except FileNotFoundError:
                    print(f"Error: File not found - {file_path}")
                except Exception as e:
                    print(f"An error occurred while processing {filename}: {e}")

        if all_first_column_values:
            value_counts = pd.Series(all_first_column_values).value_counts().sort_index()
            print(f"Value counts for the first column in {split} split:")
            print(value_counts)
        else:
            print(f"No label data found in {split} split.")
    else:
        print(f"Labels directory not found for split {split}: {labels_dir}")
```
